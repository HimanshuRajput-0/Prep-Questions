**Find the smallest number in the array:**

````python
# using sort() function:

def smallernumber (arr):
  arr.sort()

  return arr[0]

if __name__ == "__main__":

  arr = [7, 9, 5, 10, 8]
  min = smallernumber(arr)

  print(f"the smallest number in the array is {min}")
````

```
Time Complexity: O(N log N), where N is the size of the array, as we are sorting the array.

Space Complexity: O(1), as we are using a constant.
```

```OR```

````python
# using optimal approach:

def smallestnumber (arr):
  min = arr[0]

  for i in range(len(arr)):
    if min>arr[i]:
      min = arr[i]

  return min

if __name__ == "__main__":

  arr = [7, 9, 5, 10, 8]
  min = smallestnumber(arr)

  print(f"the smallest number in the array is {min}")
````

```
Time Complexity: O(N), where N is the size of the array, as we are iterating through the array once.
```

```Space Complexity: O(1), as we are using a constant.```

**Find Second Smallest and Second Largest Element in an array :**

````python
# using sort() function:

def secondsmallestandLargestnum (arr):

  arr.sort()

  secondmin = arr[1]
  secondmax = arr[-2]

  return secondmin, secondmax

if __name__ =="__main__":

  arr = [8, 10, 5, 7, 9]

  secondmin, secondmax = secondsmallestandLargestnum(arr)

  print(f"the second smallest number in this array is {secondmin}")
  print(f"the second largest number in this array is {secondmax}")
````

```
Time Complexity: O(N log N), for sorting the array.

Space Complexity: O(1), as we are using a constant amount of space for variables.
```

```OR```

````python
# Optimal Approach:

def secondsmallestnum (arr):

  smallest = secondsmallest = float('inf')

  for i in range(len(arr)):
    if arr[i]<smallest:
      secondsmallest = smallest
      smallest = arr[i]
    elif arr[i] < secondsmallest and arr[i] != smallest:
      secondsmallest = arr[i]

  return secondsmallest

def secondlargestnum(arr):

  largest = secondlargest = float('-inf')

  for i in range(len(arr)):
    if arr[i]>largest:
      secondlargest = largest
      largest = arr[i]
    elif arr[i] > secondlargest and arr[i] != largest:
      secondlargest = arr[i]

  return secondlargest


if __name__ =="__main__":
  arr = [8, 10, 5, 7, 9]

  minresult = secondsmallestnum(arr)
  maxresult = secondlargestnum(arr)

  print(f"the second smallest number is {minresult}")
  print(f"the second largest number is {maxresult}")
````

```
Time Complexity: O(N), we do two linear traversals in our array.
```

```Space Complexity: O(1), as we are using a constant.```

**Reverse a given array :**

````python
# using slicing:

def revarr (arr):

  rev = arr[::-1]

  return rev

if __name__ == "__main__":

  arr = [5,4,3,2,1]
  result = revarr(arr)

  print(f"reversed array is {result}")
````

```Time Complexity: O(n), because each element is visited once and possibly swapped once with its mirror index.```

```Space Complexity: O(1) for C++, Java, and JavaScript (in-place), but O(n) for Python slicing since it creates a new list and then assigns back (unless using two pointers).```

```OR```

````python
# using two pointers and swapping:
# Inplace algorithm:

def revarr (arr):

  p1 = 0
  p2 = len(arr)-1

  while p1 < p2:

      arr[p1], arr[p2] = arr[p2], arr[p1]
      p1 += 1
      p2 -= 1

  return arr

if __name__ == "__main__":

  arr = [1,2,3,4,5]
  result = revarr(arr)

  print(result)
````

```Time Complexity: O(n) Where n is the number of elements in the array. Each element is visited at most once due to the two-pointer approach.```

```Space Complexity: O(1) No extra space is used other than a few pointers and variables. The array is reversed in-place.```

```OR```

````python
# using another array:

def revarr (arr):

  n = len(arr)
  rev = [0]*n

  for i in range(n):
    rev[i] = arr[n -1 - i]

  return rev

if __name__ == "__main__":

  arr = [1,2,3,4,5]
  result = revarr(arr)

  print(result)
````

```Time Complexity: O(n) Each element is visited once in a loop, where n is the number of elements in the input array.```

```Space Complexity: O(n) An additional array of the same size is used to store the reversed elements.```
