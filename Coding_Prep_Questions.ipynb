**Find the smallest number in the array:**

````python
# using sort() function:

def smallernumber (arr):
  arr.sort()

  return arr[0]

if __name__ == "__main__":

  arr = [7, 9, 5, 10, 8]
  min = smallernumber(arr)

  print(f"the smallest number in the array is {min}")
````

```
Time Complexity: O(N log N), where N is the size of the array, as we are sorting the array.

Space Complexity: O(1), as we are using a constant.
```

```OR```

````python
# using optimal approach:

def smallestnumber (arr):
  min = arr[0]

  for i in range(len(arr)):
    if min>arr[i]:
      min = arr[i]

  return min

if __name__ == "__main__":

  arr = [7, 9, 5, 10, 8]
  min = smallestnumber(arr)

  print(f"the smallest number in the array is {min}")
````

```
Time Complexity: O(N), where N is the size of the array, as we are iterating through the array once.

Space Complexity: O(1), as we are using a constant.
```

**Find the largest number in the array:**

````python
# using sort() function:

def largestnumber (arr):
  arr.sort()

  return arr[-1]

if __name__ == "__main__":

  arr = [8, 10, 5, 7, 9]

  max = largestnumber(arr)
  print(f"the largest number in this array is {max}")
````
